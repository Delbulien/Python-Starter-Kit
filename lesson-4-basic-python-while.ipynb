```python
# Type conversion

# This is a string
a="tutu"

# This is a number store in a string variable
b="1234"
#To get the type of a variable you can use the command type()
print(type(b))
# You can convert it into an integer
c=int(b)
print(type(c), c)

# same for decimal
d="254.2567"
print(type(d))
e=float(d)
print(type(e), e)


# boolean to int
f=True
print(type(f))
g=int(f)
print(type(g), g)

# int to boolean
h=0
print(type(h))
i=bool(h)
print(type(i), i)

# Number to string
j=1.5E4
print(type(j))
k=str(j)
print(type(k), k)
```

```python
# Multiplicative persistence

print ("Input a positive integer")
number=int(input())

steps = 0
while number >= 10:
    product = 1
    for d in str(number):
        product *= int(d)
    number = product
    steps += 1

print("Persistance multiplicative :", steps)
print("Racine numérique multiplicative :", number)
```
