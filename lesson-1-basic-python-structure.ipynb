```python
# List
c=[1,2,3,4,5]
print (c)

c.append(6)
print (c)

print (min(c))
print (max(c))

c.insert(3,30)
print (c)

print (c.pop(3))
print(c)

c.insert(1,10)
print (c)

c.sort()
print(c)

d=["toto", "tutu", "titi"]
print (d)

d.append("tata")
print(d)

e=[[1,2],[1,2],[3,4]]
print(e)

f=["tarte", 10, [1,2,3],"flan"]
print(f)

g=f+e
print(g)

print ("First and last element: %s, %s" %(g[0], g[-1]))
print("Number of element in the list:", len(g))
```
